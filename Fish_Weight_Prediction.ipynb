## 1. Project Introduction

Fish weight prediction is a regression problem in which the weight of a
fish is predicted using its physical characteristics.

The dataset contains information about different fish species along with
their measurements such as vertical length, diagonal length, cross length,
height, and width.

In this project, the relationship between these physical characteristics
and fish weight will be studied using data analysis and machine learning
techniques.

The main machine learning technique used in this project is Linear
Regression.

## 2. Problem Statement

Determining the weight of a fish manually can require measurement and
weighing of individual fish.

The aim of this project is to develop a machine learning model that can
predict the weight of a fish based on its physical characteristics.

The project will analyze whether variables such as length, height, and
width can be used to predict fish weight accurately.

## 4. Dataset Description

The dataset contains information about different species of fish and their
physical measurements.

### Variables

- Species: Name of the fish species.
- Weight: Weight of the fish in grams. This is the target variable.
- Length1: Vertical length of the fish in centimeters.
- Length2: Diagonal length of the fish in centimeters.
- Length3: Cross length of the fish in centimeters.
- Height: Height of the fish in centimeters.
- Width: Diagonal width of the fish in centimeters.

The target variable in this project is Weight.
The remaining characteristics will be considered as predictor variables.

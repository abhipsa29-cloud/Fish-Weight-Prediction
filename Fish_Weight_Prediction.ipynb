## 1. Project Introduction

Fish weight prediction is a regression problem in which the weight of a
fish is predicted using its physical characteristics.

The dataset contains information about different fish species along with
their measurements such as vertical length, diagonal length, cross length,
height, and width.

In this project, the relationship between these physical characteristics
and fish weight will be studied using data analysis and machine learning
techniques.

The main machine learning technique used in this project is Linear
Regression.

## 2. Problem Statement

Determining the weight of a fish manually can require measurement and
weighing of individual fish.

The aim of this project is to develop a machine learning model that can
predict the weight of a fish based on its physical characteristics.

The project will analyze whether variables such as length, height, and
width can be used to predict fish weight accurately.

# DAY 5 — Exploratory Data Analysis (EDA)

Today, exploratory data analysis will be performed to understand the
distribution and characteristics of the fish dataset.

Basic statistical measures and visualizations will be used to study
the fish species and their weight.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Fish.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python
df_clean = df.copy()

print("Clean dataset ready.")
```

## 1. Statistical Summary

```python
df_clean.describe()
```

```python
print("Average fish weight:", df_clean["Weight"].mean(), "grams")
print("Minimum fish weight:", df_clean["Weight"].min(), "grams")
print("Maximum fish weight:", df_clean["Weight"].max(), "grams")
```

## 2. Fish Species Distribution

```python
species_counts = df_clean["Species"].value_counts()

print(species_counts)
```

```python
import matplotlib.pyplot as plt

species_counts.plot(kind="bar")

plt.title("Number of Fish by Species")
plt.xlabel("Fish Species")
plt.ylabel("Number of Fish")
plt.xticks(rotation=45)
plt.show()
```

## 3. Distribution of Fish Weight

```python
plt.hist(df_clean["Weight"], bins=20)

plt.title("Distribution of Fish Weight")
plt.xlabel("Weight (grams)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(df_clean["Weight"])

plt.title("Boxplot of Fish Weight")
plt.ylabel("Weight (grams)")
plt.show()
```

```python
heaviest_fish = df_clean.loc[df_clean["Weight"].idxmax()]

print("Heaviest fish record:")
print(heaviest_fish)
```

```python
lightest_fish = df_clean.loc[df_clean["Weight"].idxmin()]

print("Lightest fish record:")
print(lightest_fish)
```

## 4. Average Weight by Fish Species

```python
average_weight = df_clean.groupby("Species")["Weight"].mean()

print(average_weight)
```

```python
average_weight.plot(kind="bar")

plt.title("Average Fish Weight by Species")
plt.xlabel("Fish Species")
plt.ylabel("Average Weight (grams)")
plt.xticks(rotation=45)
plt.show()
```

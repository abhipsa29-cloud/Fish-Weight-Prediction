## 1. Project Introduction

Fish weight prediction is a regression problem in which the weight of a

height, and width.

In this project, the relationship between these physical characteristics
and fish weight will be studied using data analysis and machine learning
techniques.

The main machine learning technique used in this project is Linear
Regression.

## 2. Problem Statement

Determining the weight of a fish manually can require measurement and
weighing of individual fish.

The aim of this project is to develop a machine learning model that can
predict the weight of a fish based on its physical characteristics.

The project will analyze whether variables such as length, height, and
width can be used to predict fish weight accurately.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Fish.csv"

df = pd.read_csv(url)

df_clean = df.copy()

print("Dataset loaded successfully!")
```

```python
features = ["Length1", "Length2", "Length3", "Height", "Width"]

X = df_clean[features]
y = df_clean["Weight"]

print("Features:", features)
print("Target:", y.name)
```

```python
plt.scatter(df_clean["Length1"], df_clean["Weight"])

z = __import__("numpy").polyfit(df_clean["Length1"], df_clean["Weight"], 1)
p = __import__("numpy").poly1d(z)

plt.plot(df_clean["Length1"], p(df_clean["Length1"]))

plt.title("Length1 vs Weight with Trend Line")
plt.xlabel("Length1 (cm)")
plt.ylabel("Weight (grams)")
plt.show()
```

## 6. Correlation Comparison

```python
correlation_values = {}

for feature in features:
    correlation_values[feature] = df_clean[feature].corr(df_clean["Weight"])

correlation_table = pd.DataFrame(
    list(correlation_values.items()),
    columns=["Feature", "Correlation_with_Weight"]
)

correlation_table = correlation_table.sort_values(
    by="Correlation_with_Weight",
    ascending=False
)

correlation_table
```

```python
strongest = correlation_table.iloc[0]
weakest = correlation_table.iloc[-1]

print("Strongest relationship:")
print(strongest)

print("\nWeakest relationship:")
print(weakest)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    correlation_table["Feature"],
    correlation_table["Correlation_with_Weight"]
)

plt.title("Feature Correlation with Fish Weight")
plt.xlabel("Feature")
plt.ylabel("Correlation")
plt.xticks(rotation=45)
plt.show()
```

## 7. Interpretation

A positive correlation means that Weight generally tends to increase
when the feature increases.

A correlation closer to +1 indicates a stronger positive linear

# DAY 12 — Feature Importance and Feature Contribution Analysis

## Objective

To analyze the relationship between fish characteristics and fish weight, examine the Linear Regression coefficients, and identify which features contribute most strongly to the model's predictions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Fish.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nFirst five records:")
print(df.head())
```

## Dataset Description

The Fish dataset contains information about different fish species and their physical measurements.

The features selected for this analysis are Length1, Length2, Length3, Height, and Width. The target variable is Weight, measured in grams.

These features will be analyzed to understand their relationship with fish weight.

```python
features = ["Length1", "Length2", "Length3", "Height", "Width"]

X = df[features]
y = df["Weight"]

print("Selected features:")
print(features)

print("\nTarget variable:", y.name)

print("\nInput data shape:", X.shape)
print("Target data shape:", y.shape)
```

## Feature Selection

The five numerical features are selected because they represent different physical measurements of a fish.

- Length1: First length measurement.
- Length2: Second length measurement.
- Length3: Third length measurement.
- Height: Height measurement.
- Width: Width measurement.

Weight is selected as the target variable because the main objective of this project is to predict fish weight.

```python
correlations = df[features + ["Weight"]].corr()["Weight"].drop("Weight")

correlation_table = pd.DataFrame({
    "Feature": correlations.index,
    "Correlation with Weight": correlations.values
})

correlation_table["Absolute Correlation"] = (
    correlation_table["Correlation with Weight"].abs()
)

correlation_table = correlation_table.sort_values(
    by="Absolute Correlation",
    ascending=False
)

print("Feature Correlation Analysis:")
print(correlation_table.to_string(index=False))
```

## Understanding Feature Correlation

Correlation measures the strength and direction of a linear relationship between two variables.

- A positive correlation indicates that the variables tend to increase together.
- A negative correlation indicates that one variable tends to decrease as the other increases.
- A correlation near zero indicates a weak linear relationship.

The absolute correlation value measures the strength of the relationship without considering its direction.

Correlation does not prove causation, and it does not directly measure a feature's independent contribution to a multiple-feature model.

```python
plt.figure(figsize=(9, 5))

plt.bar(
    correlation_table["Feature"],
    correlation_table["Correlation with Weight"]
)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Fish Features")
plt.ylabel("Correlation with Weight")
plt.title("Correlation of Fish Features with Weight")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Correlation Graph Interpretation

The bar chart compares the correlation of each selected feature with fish weight.

Features with larger absolute correlation values have stronger linear relationships with weight in this dataset.

The graph helps identify promising features for prediction. However, correlation alone is not sufficient to decide which features should be removed from the model.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

#  Training the Linear Regression Model

The dataset is divided into training and testing sets.

The training set is used to fit the Linear Regression model. The model learns coefficients for the selected features and uses them to estimate fish weight.

The testing set is kept separate for evaluating the model's predictions.

```python
coefficient_table = pd.DataFrame({
    "Feature": features,
    "Coefficient": model.coef_
})

coefficient_table["Absolute Coefficient"] = (
    coefficient_table["Coefficient"].abs()
)

coefficient_table = coefficient_table.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

print("Linear Regression Coefficients:")
print(coefficient_table.to_string(index=False))

print("\nModel intercept:", model.intercept_)
```

## Interpretation of Linear Regression Coefficients

Each coefficient represents the change in predicted fish weight associated with a one-unit increase in that feature, while the other selected features are held constant.

- A positive coefficient indicates an increase in predicted weight when that feature increases, holding other features constant.
- A negative coefficient indicates a decrease in predicted weight under the same condition.
- The intercept represents the model's predicted weight when all input features are zero.

Important: The coefficient magnitudes should not be compared as a direct measure of importance without considering the different scales of the features and their correlations with one another.

```python
plt.figure(figsize=(9, 5))

plt.bar(
    coefficient_table["Feature"],
    coefficient_table["Coefficient"]
)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Fish Features")
plt.ylabel("Linear Regression Coefficient")
plt.title("Linear Regression Model Coefficients")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Coefficient Graph Interpretation

The coefficient graph shows the direction and magnitude of the fitted Linear Regression coefficients.

Positive coefficients contribute positively to predicted weight, while negative coefficients contribute negatively when other features are held constant.

Because fish length measurements may be strongly correlated with one another, the individual coefficients can be unstable or difficult to interpret independently.

```python
correlation_comparison = pd.DataFrame({
    "Feature": features,
    "Correlation with Weight": [
        df[feature].corr(df["Weight"]) for feature in features
    ],
    "Linear Regression Coefficient": model.coef_
})

print("Correlation and Coefficient Comparison:")
print(correlation_comparison.to_string(index=False))
```

##  Comparing Correlation and Model Coefficients

Correlation measures the individual linear relationship between a feature and fish weight.

A Linear Regression coefficient measures the feature's contribution to the model's prediction while accounting for the other selected features.

Therefore, the feature with the strongest correlation does not necessarily have the largest or most reliable regression coefficient.

Both analyses provide useful information about the dataset and the model.

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    model,
    X_test,
    y_test,
    scoring="r2",
    n_repeats=20,
    random_state=42
)

importance_table = pd.DataFrame({
    "Feature": features,
    "Importance Mean": result.importances_mean,
    "Importance Std": result.importances_std
})

importance_table = importance_table.sort_values(
    by="Importance Mean",
    ascending=False
)

print("Permutation Feature Importance:")
print(importance_table.to_string(index=False))
```

## Permutation Feature Importance

Permutation importance estimates how much a model depends on a feature for its predictive performance.

A larger positive importance value indicates that shuffling the feature generally reduces the model's R² score.

A value near zero suggests that shuffling the feature has little effect on the measured performance. Negative values can occur due to sampling variability or when shuffling happens to improve the score.

When features are strongly correlated, permutation importance may underestimate their individual importance because other features can provide similar information.

```python
plt.figure(figsize=(9, 5))

plt.bar(
    importance_table["Feature"],
    importance_table["Importance Mean"],
    yerr=importance_table["Importance Std"],
    capsize=4
)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Fish Features")
plt.ylabel("Mean Decrease in R²")
plt.title("Permutation Feature Importance")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Permutation Importance Graph Interpretation

The graph compares the effect of shuffling each feature on the model's R² score.

Features with larger positive importance values have a greater measured effect on predictive performance for this test set.

The error bars represent variability across the repeated permutations.

These results describe the current model and test set; they do not establish that a feature causes changes in fish weight.

# DAY 13 — Actual vs. Predicted Fish Weight Analysis

## Objective

To compare actual fish weights with predicted fish weights, calculate prediction errors, identify large errors, and evaluate the performance of the Linear Regression model through tables and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Fish.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## Dataset Preparation

The Fish dataset contains physical measurements of fish and their corresponding weights.

The features Length1, Length2, Length3, Height, and Width will be used to predict Weight.

The dataset will be divided into training and testing sets so that the model can be evaluated on records it has not used for training.

```python
features = ["Length1", "Length2", "Length3", "Height", "Width"]

X = df[features]
y = df["Weight"]

print("Selected features:", features)
print("Target variable:", y.name)
print("Total records:", len(df))
```

##  Feature Selection

The five input features represent different physical measurements of fish.

Weight is the target variable that the model aims to predict.

Using these features, the Linear Regression model estimates fish weight based on the relationships learned from the training data.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

print("Model trained successfully!")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

## Model Training

The dataset is divided into 80% training data and 20% testing data.

The Linear Regression model learns relationships between the input features and fish weight using the training data.

The testing data is reserved for evaluating predictions on unseen records.

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print("\nFirst 10 predicted weights:")
print(y_pred[:10])
```

## Generating Predictions

The trained model predicts the weight of each fish in the testing dataset.

These predicted weights will be compared with the actual weights to understand the model's performance.

```python
comparison = pd.DataFrame({
    "Actual Weight": y_test.values,
    "Predicted Weight": y_pred
})

comparison["Error"] = (
    comparison["Actual Weight"] - comparison["Predicted Weight"]
)

comparison["Absolute Error"] = comparison["Error"].abs()

print("Actual vs. Predicted Fish Weight:")
print(comparison.head(15).round(2))
```

## Actual vs. Predicted Comparison

The comparison table contains four columns:

- Actual Weight: The recorded weight of the fish.
- Predicted Weight: The weight estimated by the model.
- Error: Actual weight minus predicted weight.
- Absolute Error: The magnitude of the difference without considering its direction.

A smaller absolute error means that the predicted weight is closer to the actual weight.

```python
mean_error = comparison["Error"].mean()
mae = comparison["Absolute Error"].mean()

print("Mean Prediction Error:", round(mean_error, 2), "grams")
print("Mean Absolute Error:", round(mae, 2), "grams")
```

## Average Prediction Error

Mean Prediction Error indicates the average signed difference between actual and predicted weights.

- A positive mean error indicates that the model tends to underestimate weight on average.
- A negative mean error indicates that the model tends to overestimate weight on average.
- A value near zero indicates little average signed error, although individual errors may still be large.

Mean Absolute Error measures the average size of the prediction errors without allowing positive and negative errors to cancel each other out.

```python
largest_errors = comparison.sort_values(
    by="Absolute Error",
    ascending=False
)

print("10 Fish Predictions with the Largest Errors:")
print(largest_errors.head(10).round(2))
```

## Analysis of Large Prediction Errors

The table identifies fish records for which the model's predictions differ most from the actual weights.

Large errors may occur because of unusual fish measurements, complex relationships between features, measurement variability, or limitations of the Linear Regression model.

These records should be investigated rather than automatically removed from the dataset.

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.7,
    label="Fish predictions"
)

min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--",
    label="Perfect prediction"
)

plt.xlabel("Actual Weight (grams)")
plt.ylabel("Predicted Weight (grams)")
plt.title("Actual vs. Predicted Fish Weight")
plt.legend()
plt.tight_layout()
plt.show()
```

##  Actual vs. Predicted Graph Interpretation

Each point represents a fish in the testing dataset.

The horizontal axis shows actual weight, while the vertical axis shows predicted weight.

The dashed diagonal line represents perfect predictions.

- Points close to the line indicate more accurate predictions.
- Points above the line indicate overprediction.
- Points below the line indicate underprediction.

The overall spread of points helps us understand how closely the model's predictions match the actual weights.

```python
plt.figure(figsize=(8, 5))

plt.hist(
    comparison["Error"],
    bins=12,
    edgecolor="black"
)

plt.axvline(
    x=0,
    linestyle="--",
    label="Zero error"
)

plt.xlabel("Prediction Error (grams)")
plt.ylabel("Number of Fish")
plt.title("Distribution of Fish Weight Prediction Errors")
plt.legend()
plt.tight_layout()
plt.show()
```

##  Prediction Error Distribution

The histogram shows how prediction errors are distributed across the testing dataset.

Errors near zero represent predictions that are close to the actual weights.

Positive errors indicate underprediction, while negative errors indicate overprediction.

A wide distribution indicates that prediction errors vary considerably across fish records.

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

evaluation = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R² Score"],
    "Value": [mae, mse, rmse, r2]
})

print("Model Evaluation Results:")
print(evaluation.round(3).to_string(index=False))
```

##  Model Evaluation

The model is evaluated using four metrics:

- MAE: Average absolute prediction error, measured in grams.
- MSE: Average squared prediction error, measured in squared grams.
- RMSE: Square root of MSE, measured in grams.
- R² Score: Measures how much of the variation in actual fish weight is explained by the model relative to predicting the test-set mean.

Lower MAE, MSE, and RMSE generally indicate smaller prediction errors. A higher R² Score generally indicates better explanatory performance on the test dataset.

These metrics should be interpreted together because no single metric completely describes model performance.
